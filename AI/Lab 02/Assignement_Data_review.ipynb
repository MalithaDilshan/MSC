```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
english_stops = set(stopwords.words('english'))


def load_dataset(tweet_data):
    x_data = tweet_data['tweet']       # Reviews/Input  --> Object
    y_data = tweet_data['label']    # Sentiment/Output  --> Int

    # PRE-PROCESS REVIEW
    x_data = x_data.replace({'<.*?>': ''}, regex = True)          # remove html tag
    x_data = x_data.replace({'[^A-Za-z]': ' '}, regex = True)     # remove non alphabet
    x_data = x_data.apply(lambda review: [w for w in review.split() if w not in english_stops])  # remove stop words
    x_data = x_data.apply(lambda review: [w.lower() for w in review])   # lower case

    # y_data already encoded as 0 and 1 ( int values)

    return x_data, y_data
```

```python
X_train, y_train = load_dataset(train_data)
x_test, y_test = load_dataset(test_data)

print('Train Data ------------------------')
print('Tweets')
print(X_train, '\n')
print('Label')
print(y_train)
print('\n')
print('Test Data ------------------------')
print('Tweets')
print(x_test, '\n')
print('Label')
print(y_test)
```
